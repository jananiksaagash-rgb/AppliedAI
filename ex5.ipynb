```python
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score, average_precision_score, RocCurveDisplay, PrecisionRecallDisplay

np.random.seed(42)
tf.random.set_seed(42)

url = r"C:\Users\Janani\Downloads\creditcard.csv"
df = pd.read_csv(url)

X = df.drop("Class", axis=1)
y = df["Class"]

def preprocess_data(X_train, X_test):
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    return X_train_scaled, X_test_scaled

def compute_class_weights(y):
    neg, pos = np.bincount(y)
    total = neg + pos
    return {0: total / (2 * neg), 1: total / (2 * pos)}

def build_dnn(input_dim):
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(input_dim,)),
        tf.keras.layers.Dense(128, activation="relu"),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Dropout(0.4),
        tf.keras.layers.Dense(64, activation="relu"),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Dropout(0.3),
        tf.keras.layers.Dense(32, activation="relu"),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Dropout(0.2),
        tf.keras.layers.Dense(16, activation="relu"),
        tf.keras.layers.Dense(1, activation="sigmoid")
    ])
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
        loss="binary_crossentropy",
        metrics=[
            tf.keras.metrics.AUC(name="roc_auc"),
            tf.keras.metrics.AUC(name="pr_auc", curve="PR")
        ]
    )
    return model

def evaluate_model(model, X_test, y_test, exp_name=""):
    y_pred_prob = model.predict(X_test).ravel()
    y_pred = (y_pred_prob >= 0.5).astype(int)
    print(f"\n--- {exp_name} Metrics ---")
    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, digits=4))
    roc_auc = roc_auc_score(y_test, y_pred_prob)
    pr_auc = average_precision_score(y_test, y_pred_prob)
    print(f"ROC-AUC: {roc_auc:.4f}")
    print(f"PR-AUC: {pr_auc:.4f}")
    RocCurveDisplay.from_predictions(y_test, y_pred_prob)
    PrecisionRecallDisplay.from_predictions(y_test, y_pred_prob)
    plt.show()
    return roc_auc, pr_auc

print("\n--- Experiment 1: 80-20 Train-Test Split ---")
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
X_train, X_test = preprocess_data(X_train, X_test)
model = build_dnn(X_train.shape[1])
class_weights = compute_class_weights(y_train)
history = model.fit(X_train, y_train, epochs=20, batch_size=2048, class_weight=class_weights, verbose=1)
plt.plot(history.history["loss"], label="Train Loss")
plt.title("Loss Curve (Experiment 1)")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
evaluate_model(model, X_test, y_test, exp_name="Experiment 1")

print("\n--- Experiment 2: 70-30 Train-Test Split ---")
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y, random_state=42)
X_train, X_test = preprocess_data(X_train, X_test)
model = build_dnn(X_train.shape[1])
class_weights = compute_class_weights(y_train)
history = model.fit(X_train, y_train, epochs=20, batch_size=2048, class_weight=class_weights, verbose=1)
plt.plot(history.history["loss"], label="Train Loss")
plt.title("Loss Curve (Experiment 2)")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
evaluate_model(model, X_test, y_test, exp_name="Experiment 2")

print("\n--- Experiment 3: Train-Validation-Test Split with Early Stopping ---")
X_temp, X_test, y_temp, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.25, stratify=y_temp, random_state=42)
X_train, X_val = preprocess_data(X_train, X_val)
X_test = StandardScaler().fit_transform(X_test)
model = build_dnn(X_train.shape[1])
class_weights = compute_class_weights(y_train)
early_stop = tf.keras.callbacks.EarlyStopping(monitor="val_pr_auc", mode="max", patience=5, restore_best_weights=True)
history = model.fit(X_train, y_train, validation_data=(X_val, y_val), epochs=50, batch_size=2048, class_weight=class_weights, callbacks=[early_stop], verbose=1)
plt.plot(history.history["loss"], label="Train Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Loss Curves (Experiment 3)")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
evaluate_model(model, X_test, y_test, exp_name="Experiment 3")

print("\n--- Experiment 4: Reduced Test Size with Early Stopping ---")
X_temp, X_test, y_temp, y_test = train_test_split(X, y, test_size=0.15, stratify=y, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.1765, stratify=y_temp, random_state=42)
X_train, X_val = preprocess_data(X_train, X_val)
X_test = StandardScaler().fit_transform(X_test)
model = build_dnn(X_train.shape[1])
class_weights = compute_class_weights(y_train)
early_stop = tf.keras.callbacks.EarlyStopping(monitor="val_pr_auc", mode="max", patience=5, restore_best_weights=True)
history = model.fit(X_train, y_train, validation_data=(X_val, y_val), epochs=50, batch_size=2048, class_weight=class_weights, callbacks=[early_stop], verbose=1)
plt.plot(history.history["loss"], label="Train Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Loss Curves (Experiment 4)")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
evaluate_model(model, X_test, y_test, exp_name="Experiment 4")

print("\n--- Experiment 5: Stratified 5-Fold Cross Validation ---")
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
roc_scores = []
pr_scores = []

for fold, (train_idx, test_idx) in enumerate(skf.split(X, y), 1):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    X_train, X_test = preprocess_data(X_train, X_test)
    model = build_dnn(X_train.shape[1])
    class_weights = compute_class_weights(y_train)
    history = model.fit(X_train, y_train, epochs=20, batch_size=2048, class_weight=class_weights, verbose=0)
    plt.plot(history.history["loss"], label=f"Fold {fold} Train Loss")
    plt.title(f"Loss Curve (Experiment 5 - Fold {fold})")
    plt.xlabel("Epochs")
    plt.ylabel("Loss")
    plt.legend()
    plt.show()
    y_pred_prob = model.predict(X_test).ravel()
    roc = roc_auc_score(y_test, y_pred_prob)
    pr = average_precision_score(y_test, y_pred_prob)
    print(f"\nFold {fold} Metrics:")
    print(f"ROC-AUC: {roc:.4f}, PR-AUC: {pr:.4f}")
    RocCurveDisplay.from_predictions(y_test, y_pred_prob)
    PrecisionRecallDisplay.from_predictions(y_test, y_pred_prob)
    plt.show()
    roc_scores.append(roc)
    pr_scores.append(pr)

print("\nExperiment 5 - Cross Validation Results")
print(f"Mean ROC-AUC: {np.mean(roc_scores):.4f} ± {np.std(roc_scores):.4f}")
print(f"Mean PR-AUC: {np.mean(pr_scores):.4f} ± {np.std(pr_scores):.4f}")
```
